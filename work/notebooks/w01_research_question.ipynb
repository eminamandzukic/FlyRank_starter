## 2. The question: decision, action, cost of a wrong call

### Research question

Can observable search, engagement, and content signals help rank which pages should be reviewed first for possible refresh, expansion, protection, pruning, or monitoring?

### Decision

The decision is which pages should receive reviewer attention first when the review team has limited time and cannot inspect every page.

### Action

A content or SEO reviewer would examine the highest-ranked pages and decide whether each page should be refreshed, expanded, protected, pruned, or simply monitored. The ranking itself would not automatically trigger a content change.

### Cost of a wrong call

A false positive occurs when a page is ranked highly even though it does not meaningfully need attention. This can waste reviewer time and reduce the capacity available for more important pages.

A false negative occurs when a page with a meaningful decline or opportunity is ranked too low and is therefore not reviewed. This may allow a potentially valuable issue to remain unnoticed.

Because reviewer capacity is limited, the ranking should prioritize useful candidates near the top of the queue rather than only maximize overall classification accuracy.


## 3. Quick look at the data (2-3 real numbers)

It can be observed here that the starter dataset contains 30,000 content items from 32 pseudonymized clients. Of these pages, 16,262 (54.2%) are categorized as having a downward recent impression trend. The median page also received 731 search impressions during the trailing 90-day period.

These numbers make the Refresh / Content Opportunity Scoring lane worth investigating because downward movement is common in the starter data and many pages have enough search exposure for prioritization to matter. At the same time, a downward trend alone should not be treated as evidence that a page requires a refresh, so the purpose of the project is to investigate whether multiple observable signals can improve which pages are selected for human review.


```python
import pandas as pd

df = pd.read_csv("../../data/raw/content_refresh_anonymized.csv")

n_pages = len(df)
n_clients = df["client_id"].nunique()

n_declining = (df["trend_direction"] == "down").sum()
declining_pct = 100 * n_declining / n_pages

median_impressions = df["impressions_90d"].median()

print(f"Content items: {n_pages:,}")
print(f"Clients: {n_clients}")
print(f"Pages with a downward trend: {n_declining:,} ({declining_pct:.1f}%)")
print(f"Median 90-day impressions per page: {median_impressions:,.0f}")
```
